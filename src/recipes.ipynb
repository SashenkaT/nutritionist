```python
import pandas as pd

df = pd.read_csv('epi_r.csv')
print("Размер датасета:", df.shape)
df.head()
```

```python
df = df.dropna()
df.info()
df['#cakeweek'].value_counts()
```

```python
df.describe()
df = df[df['calories'] <= 5000]
df = df[df['sodium'] <= 10000]
df.describe()

# Сохраняем названия рецептов
recipe_titles = df['title'].copy()

df = df.drop(['title', 'calories', 'protein', 'fat', 'sodium'], axis=1)
df.head()
```

```python
from sklearn.model_selection import train_test_split

y = df['rating']
X = df.drop('rating', axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
# Линейная регрессия
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

model1 = LinearRegression()
model1.fit(X_train, y_train)
predict = model1.predict(X_test)
res = np.sqrt(mean_squared_error(y_test, predict))

print(res)
```

```python
# Наивный регрессор
from sklearn.dummy import DummyRegressor

model2 = DummyRegressor(strategy="mean")
model2.fit(X_train, y_train)
predict = model2.predict(X_test)
res = np.sqrt(mean_squared_error(y_test, predict))

print(res)
```

```python
# Случайный лес
from sklearn.ensemble import RandomForestRegressor

model3 = RandomForestRegressor(n_estimators=150, max_depth=20, random_state=1)
model3.fit(X_train, y_train)
predict = model3.predict(X_test)
res = np.sqrt(mean_squared_error(y_test, predict))

print(res)
```

```python
# Случайный лес
from sklearn.ensemble import RandomForestRegressor

model3 = RandomForestRegressor(random_state=1)
model3.fit(X_train, y_train)
predict = model3.predict(X_test)
res = np.sqrt(mean_squared_error(y_test, predict))

print(res)
```

```python
# Классификация
import numpy as np
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

# 1. Округляем рейтинги до целого числа, создавая классы для классификации
y_train_class = np.round(y_train)
y_test_class = np.round(y_test)

# 2. Создаем наивный классификатор (стратегия - выдавать самый частый рейтинг)
dummy_clf = DummyClassifier(strategy="most_frequent")
dummy_clf.fit(X_train, y_train_class)
dummy_preds = dummy_clf.predict(X_test)

# 3. Считаем новую метрику - accuracy (доля правильных ответов)
dummy_accuracy = accuracy_score(y_test_class, dummy_preds)
print(dummy_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier

model_clf1 = RandomForestClassifier(random_state=1)
model_clf1.fit(X_train, y_train_class)
predict_clf = model_clf1.predict(X_test)

res_clf = accuracy_score(y_test_class, predict_clf)
print(res_clf)
```

```python
# Разбиваем на 3 класса
cond_train = [y_train_class <= 1, (y_train_class >= 2) & (y_train_class <= 3), y_train_class >= 4]
cond_test = [y_test_class <= 1, (y_test_class >= 2) & (y_test_class <= 3), y_test_class >= 4]
choices = ["bad", "so-so", "excellent"]


y_train_new = np.select(cond_train, choices, default="unknown")
y_test_new = np.select(cond_test, choices, default="unknown")

# Базовая модель
model2 = DummyClassifier(strategy="most_frequent")
model2.fit(X_train, y_train_new)
predict = model2.predict(X_test)

res = accuracy_score(y_test_new, predict)
print(res)
```

```python
from sklearn.ensemble import RandomForestClassifier

model3 = RandomForestClassifier(random_state=1)
model3.fit(X_train, y_train_new)
predict = model3.predict(X_test)

res = accuracy_score(y_test_new, predict)
print(res)
```

```python
from sklearn.metrics import classification_report

# Выводим подробный отчет по всем классам (bad, so-so, excellent)
report = classification_report(y_test_new, predict)
print(report)
```

**Метрика:**
accuracy мимо  (почти всё excellent)
Главный риск — выдать плохой рецепт за топ 
Поэтому берем Precision

```python
from sklearn.metrics import precision_score

res_precision = precision_score(y_test_new, predict, average='macro')
print(res_precision)
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

# Обучаем градиентный бустинг
model_gb = HistGradientBoostingClassifier(random_state=1)
model_gb.fit(X_train, y_train_new)
predict_gb = model_gb.predict(X_test)

# Считаем Precision
res_precision_gb = precision_score(y_test_new, predict_gb, average='macro')
print(res_precision_gb)
```

Классификация лучше регрессии, так как нам важно не советовать плохие рецепты под видом хороших. 
Лучшая модель — RandomForestClassifier.

```python
import pandas as pd

# нормы
dv_p, dv_f, dv_c, dv_s = 50, 78, 275, 2300

df_nutr = pd.read_csv('DB/nutrient.csv')
df_vals = pd.read_csv('DB/fndds_ingredient_nutrient_value.csv')

# склеиваем по ID
df_full = df_vals.merge(df_nutr, left_on='Nutrient code', right_on='nutrient_nbr')

# оставляем нужное и в нижний регистр
df_full = df_full[['Ingredient description', 'name', 'Nutrient value']]
df_full['Ingredient description'] = df_full['Ingredient description'].str.lower()

ingredients = X_train.columns.to_list()
results = []

for item in ingredients:
    # ищем совпадения
    match = df_full[df_full['Ingredient description'].str.contains(item.lower(), regex=False, na=False)]
    
    if not match.empty:
        # берем первый найденный
        first_product = match['Ingredient description'].iloc[0]
        prod_data = match[match['Ingredient description'] == first_product]
        
        p, f, c, s = 0, 0, 0, 0
        
        # достаем цифры
        for _, row in prod_data.iterrows():
            n = str(row['name']).lower()
            val = row['Nutrient value']
            if 'protein' in n: p = val
            elif 'lipid' in n or 'fat' in n: f = val
            elif 'carbohydrate' in n: c = val
            elif 'sodium' in n: s = val
        results.append({
            'ingredient': item,
            'protein_%': round((p / dv_p) * 100, 1),
            'fat_%': round((f / dv_f) * 100, 1),
            'carbs_%': round((c / dv_c) * 100, 1),
            'sodium_%': round((s / dv_s) * 100, 1)
        })
    else:
        # если нет - ставим нули
        results.append({
            'ingredient': item, 'protein_%': 0, 'fat_%': 0, 'carbs_%': 0, 'sodium_%': 0
        })

# сохраняем
df_res = pd.DataFrame(results)
df_res.to_csv('nutrients.csv', index=False)
# display(df_res.head())
# Показываю 10 строк, где белок больше 0
display(df_res[df_res['protein_%'] > 0].head(10))
cols = [c for c in df.columns if any(x in c.lower() for x in ['title', 'url', 'link', 'name'])]
print("Подходящие колонки:", cols)
# display(df.head(2))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# берем первый рецепт
target_index = 0 
target_title = recipe_titles.iloc[target_index]
print("Ищем похожее на:", target_title)
print("-" * 30)

# считаем сходство
similarities = cosine_similarity(df.iloc[[target_index]], df)[0]

# делаем таблицу
recomm = pd.DataFrame({
    'title': recipe_titles,
    'similarity': similarities
})

# берем топ-5 (без первого)
top_5 = recomm.sort_values(by='similarity', ascending=False).iloc[1:6]

# выводим ссылки
for i, row in top_5.iterrows():
    title = str(row['title'])
    score = round(row['similarity'] * 100, 1)
    
    # делаем ссылку
    query = title.replace(' ', '+')
    link = f"https://www.google.com/search?q={query}+recipe"
    
    print("Рецепт:", title)
    print("Совпадение:", score, "%")
    print("Ссылка:", link)
```
